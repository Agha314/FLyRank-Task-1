## 4. The unit of analysis, as a real dataframe

**One row in our data equals one specific web page.**

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/content_refresh_anonymized.csv')

# Show what one row looks like
print("--- What is one row? (First 3 Rows) ---")
display(df.head(3))

# Create our proxy target: Flag as '1' if clicks dropped compared to last month,
# but people are still searching for the topic.
df['is_declining_proxy'] = (
    (df['clicks_last_30d'] < df['clicks_prev_30d']) &
    (df['search_volume'] > 0)
).astype(int)

print("\n--- A quick look at our new Target Column ---")
columns_to_show = ['content_id', 'search_volume', 'clicks_prev_30d', 'clicks_last_30d', 'is_declining_proxy']
display(df[columns_to_show].head())
```

## 5. Why ML beats a fixed rule here

**A simple human rule is too rigid. Machine learning sees the big picture.**

**The Simple Reason:** If a human writes a rule like "flag any page that lost 20% of its traffic," that rule is blind to other important details. What if the page is brand new? What if the competition is really high? Machine learning is better because it looks at all these moving parts (age, traffic, competition) at the same time to make a much smarter recommendation.
